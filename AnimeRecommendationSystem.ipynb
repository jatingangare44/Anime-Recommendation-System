```python
import pandas as pd
```

```python
anime = pd.read_csv(r"data\anime.csv")
```

```python
anime
```

```python
anime = anime[["uid","title","synopsis","genre"]]
```

```python
anime
```

```python
anime.isnull().sum()
```

```python
def remove_space(L):
    L1 = []
    for i in L:
        L1.append(i.replace(" ",""))
    return L1
```

```python
import ast
anime["genre"] = anime["genre"].apply(ast.literal_eval)
```

anime['title'] = anime['title'].apply(remove_space)
anime['synopsis'] = anime['synopsis'].apply(remove_space)
anime['genre'] = anime['genre'].apply(remove_space)

```python
anime.head()
```

```python
anime['synopsis'] = anime['synopsis'].apply(lambda x: x.split() if isinstance(x, str) else x)
anime.sample(4)
```

```python
anime["tags"] = anime["synopsis"] + anime["genre"]
```

```python
anime.head()
```

```python
new_df = anime[['uid','title','tags']]
```

```python
new_df.head()
```

```python
new_df.iloc[0]['tags']
```

```python
# Converting list to str
new_df['tags'] = new_df['tags'].apply(lambda x: " ".join(x))
new_df.head()
```

```python
new_df.iloc[1]['tags']
```

```python
anime.iloc[1]['synopsis']
```

```python
new_df['tags'] = new_df['tags'].apply(lambda x:x.lower())
```

```python
new_df.head()
```

```python
import nltk
from nltk.stem import PorterStemmer
```

```python
ps = PorterStemmer()
```

```python
def stems(text):
    T = []
    
    for i in text.split():
        T.append(ps.stem(i))
    
    return " ".join(T)
```

```python
new_df['tags'] = new_df['tags'].apply(stems)
```

```python
new_df.iloc[0]['tags']
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=5000,stop_words='english')
```

```python
vector = cv.fit_transform(new_df['tags']).toarray()
```

```python
vector[0]
```

```python
vector.shape
```

```python
len(cv.get_feature_names_out())
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vector)
```

```python
similarity
```

```python
new_df[new_df['title'] == 'Made in Abyss'].index[0]
```

```python
def recommend(anime):
    index = new_df[new_df['title'] == anime].index[0]
    distances = sorted(list(enumerate(similarity[index])),reverse=True,key = lambda x: x[1])
    for i in distances[1:6]:
        print(new_df.iloc[i[0]].title)
```

```python
recommend('Made in Abyss')
```
